### Gráficos de Dispersión:
* scatter(x, y): Dibuja un diagrama de puntos con las coordenadas de la lista x en el eje X y las coordenadas de la lista y en el eje Y.

Utilidad: Se utilizan para mostrar la relación entre dos variables numéricas. Cada punto en el gráfico representa una observación.

Cuándo usarlo:
- Para analizar correlaciones entre dos variables.
- Identificar patrones o tendencias en conjuntos de datos grandes.

Ejemplo: Relación entre la altura y el peso de las personas.

```python
fig, ax = plt.subplots()
#Crea una figura en la que se van a representar los ejes ax
ax.scatter(x = [1, 2, 3], y = [3, 2, 1])
# Guardar el gráfico en formato png
#plt.savefig('diagrama-dispersion.png')
# Mostrar el gráfico
plt.show()
```

### Gráficos de Línea:
* plot(x, y): Dibuja un polígono con los vértices dados por las coordenadas de la lista x en el eje X y las coordenadas de la lista y en el eje Y.

Utilidad: Se utilizan para mostrar tendencias a lo largo de un periodo de tiempo o un conjunto continuo de datos.

Cuándo usarlo:
- Para analizar cómo cambian los valores a lo largo del tiempo.
- Comparar dos o más series temporales.

Ejemplo: Evolución de las ventas por mes.

```python
fig, ax = plt.subplots()
ax.plot([1, 2, 3, 4], [1, 2, 0, 0.5])
plt.show()
```

### Diagrama de áreas:
* fill_between(x, y): Dibuja el area bajo el polígono con los vértices dados por las coordenadas de la lista x en el eje X y las coordenadas de la lista y en el eje Y.

Utilidad: Similar a las gráficas de líneas, pero las áreas debajo de las líneas están rellenas. Se usa para mostrar tanto la tendencia como la magnitud de la suma.

Cuándo usarlo:
- Cuando necesitas mostrar la contribución de diferentes categorías a lo largo del tiempo.
- Para visualizar la proporción que cada serie contribuye a un total.

Ejemplo: Evolución de las ventas en diferentes regiones a lo largo del tiempo, mostrando las proporciones.

```python
fig, ax = plt.subplots()
ax.fill_between([1, 2, 3, 4], [1, 2, 0, 0.5])
plt.show()
```

### Diagrama de barras verticales:
* bar(x, y): Dibuja un diagrama de barras verticales donde x es una lista con la posición de las barras en el eje X, e y es una lista con la altura de las barras en el eje Y.

Utilidad: También para comparar categorías, pero las barras son verticales. Este tipo es más común para datos con categorías discretas.

Cuándo usarlo:
- Comparar datos en diferentes categorías.

Ejemplo: Cantidad de productos vendidos por tienda.

```python
fig, ax = plt.subplots()
ax.bar([1, 2, 3], [3, 2, 1])
plt.show()
```

### Diagrama de barras horizontales:
* barh(x, y): Dibuja un diagrama de barras horizontales donde x es una lista con la posición de las barras en el eje Y, e y es una lista con la longitud de las barras en el eje X.

Utilidad: Se utilizan para comparar categorías. Son útiles cuando las etiquetas de las categorías son largas o los valores están cerca unos de otros.

Cuándo usarlo:
- Comparar cantidades de categorías distintas.
- Mejor visualización cuando las etiquetas son más largas.

Ejemplo: Comparar el número de ventas por país.

```python
fig, ax = plt.subplots()
ax.barh([1, 2, 3], [3, 2, 1])
plt.show()
```

### Histograma:
* hist(x, bins): Dibuja un histograma con las frecuencias resultantes de agrupar los datos de la lista x en las clases definidas por la lista bins.

Utilidad: Se usan para mostrar la distribución de un conjunto de datos continuos dividiendo los datos en intervalos (bins). Permiten visualizar la frecuencia de ocurrencias dentro de esos intervalos.

Cuándo usarlo:
- Para analizar la distribución de una variable continua.
- Entender la dispersión y variabilidad de los datos.

Ejemplo: Distribución de edades en un grupo de personas.

```python
x = np.random.normal(5, 1.5, size=1000)
```

```python
fig, ax= plt.subplots()
x = np.random.normal(5, 1.5, size=1000)
ax.hist(x, np.arange(0, 11))
plt.show()
```

### Diagrama de sectores:
* pie(x): Dibuja un diagrama de sectores con las frecuencias de la lista x.
import matplotlib.pyplot as plt

Utilidad: Se usa para mostrar la proporción o el porcentaje que representa cada categoría respecto a un total. Cada "sector" o "porción" del círculo representa una parte del todo.

Cuándo usarlo:
- Para visualizar cómo se distribuyen partes de un todo.
- Es útil cuando deseas enfatizar la comparación de proporciones entre diferentes categorías.
- Es más efectivo cuando tienes pocas categorías (idealmente menos de 5-6), ya que demasiadas porciones pueden dificultar la lectura.

Ejemplo: Distribución del mercado entre diferentes competidores, porcentaje de uso de distintas redes sociales, o repartición de presupuestos.

```python
grupos = ["G1", "G2", "G3", "G4", "G5"]
valores = [12, 22, 16, 38, 12]

fig, ax = plt.subplots()
ax.pie(valores, labels=grupos)
plt.show()
```

### Boxplot:
* boxplot(x): Dibuja un diagrama de caja y bigotes con los datos de la lista x.

Utilidad:
El boxplot es útil para mostrar la distribución de los datos y resaltar los valores atípicos (outliers), los cuartiles y la mediana. Proporciona una visión clara de la dispersión y simetría de los datos sin asumir una distribución específica.

Cuándo usar:
- Comparar distribuciones de diferentes grupos o variables.
- Identificar valores atípicos.
- Entender la variabilidad y simetría de los datos.

Imagina que trabajas en análisis de rendimiento académico y quieres comparar las notas de los estudiantes en diferentes asignaturas (Matemáticas, Ciencia, Inglés). Con un boxplot, puedes ver la distribución de las notas, identificar estudiantes que tienen rendimientos extremadamente bajos o altos (outliers) y comparar la dispersión de las notas entre asignaturas.

```python
fig, ax = plt.subplots()
ax.boxplot([1, 2, 1, 2, 3, 4, 3, 3, 5, 7])
plt.show()
```

### Mapas de calor:
* imshow(x): Dibuja un mapa de color a partir de una matriz (array bidimensiona) x.

Utilidad:
Un mapa de calor (heatmap) es útil para visualizar matrices de datos o correlaciones entre variables. Los valores son representados por colores, facilitando la identificación de patrones, agrupaciones o anomalías.

Cuándo usar:
- Visualizar correlaciones entre variables en una matriz de correlación.
- Visualizar valores de una matriz (p. ej., una tabla de confusión).
- Identificar patrones en datos bidimensionales.

En el análisis de ventas en una tienda online, puedes utilizar un mapa de calor para visualizar la correlación entre diferentes variables (por ejemplo, el precio del producto, el número de unidades vendidas, la satisfacción del cliente). Esto ayuda a identificar qué variables están relacionadas entre sí y puede ayudar a optimizar precios o promociones.

```python
from sklearn import datasets
import pandas as pd
# Cargamos los datos provenientes de una librería
iris = datasets.load_iris()

# Guardamos los datos de la keys "data" en un dataframe de pandas.
datos = pd.DataFrame(iris["data"])

# Asignamos al dataframe el nombre de las columnas.
datos.columns = iris["feature_names"]

# Insertamos como columna en el dataframe los valores de la key "target".
datos["species"] = iris["target"]

iris.feature_names
```

```python
variables_numericas = datos.drop(columns=['species'])

# Calcular la matriz de correlación
correlation_matrix = variables_numericas.corr()
```

```python
correlation_matrix
```

```python
fig, ax = plt.subplots()
x = np.random.random((16, 16))
ax.imshow(correlation_matrix)
plt.show()
```

### Marcadores de un plot

```python
df=pd.DataFrame(
    {
        'Días':['L', 'M', 'X', 'J', 'V', 'S', 'D'],
        'Madrid':[28.5, 30.5, 31, 30, 28, 27.5, 30.5],
        'Barcelona':[24.5, 25.5, 26.5, 25, 26.5, 24.5, 25]
    }
)
```

```python
fig, ax = plt.subplots()
ax.plot(df["Días"], df['Madrid'], marker = '^')
ax.plot(df["Días"], df['Barcelona'], marker = 'o')
plt.show()
```

### Formato de la líneas

```python
fig, ax = plt.subplots()
ax.plot(df["Días"], df['Madrid'], linestyle = 'dashed')
ax.plot(df["Días"], df['Barcelona'], linestyle = 'dotted')
plt.show()
```

### Títulos

```python
fig, ax = plt.subplots()
ax.plot(df["Días"], df['Madrid'])
ax.plot(df["Días"], df['Barcelona'])
ax.set_title('Evolución de la temperatura diaria', loc = "left", fontdict = {'fontsize':14, 'fontweight':'bold', 'color':'tab:blue'})
plt.show()
```

* ax.set_title(titulo, loc=alineacion, fontdict=fuente) : Añade un título con el contenido de la cadena titulo a los ejes ax. El parámetro loc indica la alineación del título, que puede ser 'left' (izquierda), 'center' (centro) o 'right' (derecha), y el parámetro fontdict indica mediante un diccionario las características de la fuente (la el tamaño fontisize, el grosor fontweight o el color color).

### Ejes

```python
fig, ax = plt.subplots()
ax.plot(df["Días"], df['Madrid'])
ax.plot(df["Días"], df['Barcelona'])
ax.set_xlabel("Días", fontdict = {'fontsize':14, 'fontweight':'bold', 'color':'tab:blue'})
ax.set_ylabel("Temperatura ºC")
ax.set_ylim([20,35])
ax.set_yticks(range(20, 35))
plt.show()
```

* ax.set_xlabel(titulo) : Añade un título con el contenido de la cadena titulo al eje x de ax. Se puede personalizar la alineación y la fuente con los mismos parámetros que para el título principal.
* ax.set_ylabel(titulo) : Añade un título con el contenido de la cadena titulo al eje y de ax. Se puede personalizar la alineación y la fuente con los mismos parámetros que para el título principal.
* ax.set_xlim([limite-inferior, limite-superior]) : Establece los límites que se muestran en el eje x de ax.
* ax.set_ylim([limite-inferior, limite-superior]) : Establece los límites que se muestran en el eje y de ax.
* ax.set_xticks(marcas) : Dibuja marcas en el eje x de ax en las posiciones indicadas en la lista marcas.
* ax.set_yticks(marcas) : Dibuja marcas en el eje y de ax en las posiciones indicadas en la lista marcas.
* ax.set_xscale(escala) : Establece la escala del eje x de ax, donde el parámetro escala puede ser 'linear' (lineal) o 'log' (logarítmica).
* ax.set_yscale(escala) : Establece la escala del eje y de ax, donde el parámetro escala puede ser 'linear' (lineal) o 'log' (logarítmica).

### Leyenda

```python
fig, ax = plt.subplots()
ax.plot(df["Días"], df['Madrid'], label = 'Madrid')
ax.plot(df["Días"], df['Barcelona'], label = 'Barcelona')
ax.legend(loc = 'upper right')
plt.show()
```

* ax.legend(leyendas, loc = posición) : Dibuja un leyenda en los ejes ax con los nombres indicados en la lista leyendas. El parámetro loc indica la posición en la que se dibuja la leyenda y puede ser 'upper left' (arriba izquierda), 'upper center' (arriba centro), 'upper right' (arriba derecha), 'center left' (centro izquierda), 'center' (centro), 'center right' (centro derecha), 'lower left' (abajo izquierda), 'lower center' (abajo centro), 'lower right' (abajo derecha). Se puede omitir la lista leyendas si se indica la leyenda de cada serie en la función que la dibuja mediante el parámetro label.

### Gráficos de Series Temporales:

```python
import pandas as pd
import numpy as np

# Crear un rango de fechas desde 2021-01-01 hasta 2023-12-31
fechas = pd.date_range(start="2021-01-01", end="2023-12-31", freq="D")

# Generar datos aleatorios para las columnas "dinero_ganado" y "compras_realizadas"
np.random.seed(42)
dinero_ganado = np.random.randint(low=100, high=1000, size=len(fechas))
compras_realizadas = np.random.randint(low=1, high=10, size=len(fechas))

# Crear el DataFrame
df = pd.DataFrame({
    "fecha": fechas,
    "dinero_ganado": dinero_ganado,
    "compras_realizadas": compras_realizadas
})
```

```python
import matplotlib.pyplot as plt

# Tu código existente para crear el gráfico
fig, ax1 = plt.subplots()
line1, = ax1.plot(df_1["fecha"], df_1["dinero_ganado"], label='Serie 1', color='b')
ax1.set_xlabel('Eje X')
ax1.set_ylabel('Serie 1', color='b')
ax1.tick_params(axis='y', labelcolor='b')
plt.xticks(rotation=30)
plt.show()
```
