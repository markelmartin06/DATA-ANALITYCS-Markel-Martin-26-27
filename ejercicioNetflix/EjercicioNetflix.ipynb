## **Ejercicio 10**
¿Qué valores de la columna rating parecen estar en la categoría incorrecta y por qué crees que se han producido esos errores?

```python
ej10 = netflix[netflix["rating"].str.contains("min", na=False)][["title", "director", "rating", "duration"]]
display(ej10)
```

**Respuesta:** los valores `74 min`, `84 min` y `66 min` no son clasificaciones por edad, son duraciones.
Corresponden a tres especiales de Louis C.K. (`Louis C.K. 2017`, `Louis C.K.: Hilarious` y `Louis C.K.: Live at the Comedy Store`), que además tienen la columna `duration` vacía.
Lo más probable es que al cargar o exportar los datos la duración se desplazara una columna a la izquierda (se quedó en `rating` en vez de en `duration`), por un error de captura o porque faltaba el rating original.

## **Ejercicio 11**
¿Cuántas películas tienen una duración entre 90 y 100 minutos?

```python
peliculas = netflix[netflix["type"] == "Movie"].copy()
peliculas["minutos"] = peliculas["duration"].str.extract(r"(\d+)").astype(float)
ej11 = peliculas[peliculas["minutos"].between(90, 100)]
print("Películas entre 90 y 100 minutos:", len(ej11))
```

## **Ejercicio 12**

```python
ej12 = netflix[(netflix["type"] == "Movie") & (netflix["duration"].str.contains("Season", na=False))]
print("Películas con 'Season' en duration:", len(ej12))
```

## **Ejercicio 23**
¿Qué directores tienen repartos que se repiten más de una vez? (agrupando por director y cast)

```python
repartos = netflix.groupby(["director", "cast"])["title"].count()
ej23 = repartos[repartos > 1].sort_values(ascending=False)
display(ej23)
```
