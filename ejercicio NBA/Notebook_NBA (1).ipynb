```python
datos = nba[(nba["Season"] == "2014 - 2015") & (nba["GP"] >= 50)].copy()

datos["PPG"] = datos["PTS"] / datos["GP"]

datos.sort_values("PPG", ascending=False)[["Player", "GP", "PTS", "PPG"]].head(1)
```

```python
datos1 = nba[(nba["Season"] == "2013 - 2014") & (nba["3PA"] >= 100)].copy()

datos1["PAT"] = datos1["3PM"] / datos1["3PA"] * 100

datos1.loc[datos1["PAT"].idxmax()][["Player", "3PA", "3PM", "PAT"]]
```
