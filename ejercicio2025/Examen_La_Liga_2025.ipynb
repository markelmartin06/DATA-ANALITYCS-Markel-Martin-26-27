### Pregunta 1 — 1 punto

¿Qué jugadores del Athletic Club han marcado más de 3 goles?

Muestra únicamente el **nombre del jugador, equipo, goles y asistencias**.

```python
import numpy as np
import pandas as pd

liga = pd.read_csv("laliga_player_stats_english (1).csv")
datos1 = liga[(liga["Team"] == "Athletic Club") & (liga["Goals scored"] > 3)][["Name", "Goals scored", "Assists"]]
```

### Pregunta 2 — 1 punto

¿Cuáles fueron los **5 jugadores que menos minutos disputaron** entre aquellos que marcaron al menos un gol?

Muestra su **nombre, equipo, minutos y goles**, ordenados de menor a mayor número de minutos.

```python
datos2 = liga[liga["Goals scored"] > 1].sort_values(by="Minutes played", ascending=True)[["Name", "Team", "Goals scored", "Minutes played"]].tail(5)
```

### Pregunta 4 — 1,5 puntos

¿Qué equipos acumularon más tarjetas amarillas entre todos sus jugadores?

Muestra los **5 equipos con mayor número total de tarjetas amarillas**, ordenados de mayor a menor.

```python
datos4 = liga.groupby("Team")["Yellow Cards"].sum()
datos4.sort_values(ascending=False).head()
```

### Pregunta 5 — 1,5 puntos

¿Qué equipo tuvo la **mayor media de goles por jugador** y cuál tuvo la **menor media de goles por jugador**?

Muestra ambos equipos y sus respectivas medias.

```python
datos5 = liga.groupby("Team")["Goals scored"].mean().sort_values(ascending=False)
display(datos5.head(1))
display(datos5.tail(1))
```

### Pregunta 6 — 2 puntos

¿Qué jugadores tienen la mejor media de goles por partido?

Utiliza las columnas `Goals` y `Matches` para crear una nueva columna llamada `Goles_por_partido`.

Muestra los **5 jugadores con mayor media de goles por partido**, indicando su nombre, equipo, partidos jugados, goles y la nueva columna, ordenados de mayor a menor.

```python
liga["Goles_por_partido"] = liga["Goals scored"] / liga["Games played"]
display(liga[["Name", "Team", "Goles_por_partido"]].sort_values(by="Goles_por_partido", ascending=False).head(5))
```
