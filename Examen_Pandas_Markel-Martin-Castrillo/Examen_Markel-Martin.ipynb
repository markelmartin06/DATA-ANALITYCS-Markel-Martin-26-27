```python
import pandas as pd
import numpy as np

df = pd.read_csv("athlete_events.csv")
df.columns
```

```python
#Pregunta 1
res1 = df[(df["NOC"] == "CUB") & (df["Sport"] == "Boxing") & (df["Year"] == 1980) & (df["City"] == "Moskva")][["Name", "Age", "Weight"]]
res1
```

```python
#Pregunta 2
res2 = df[(df["Season"] == "Winter") & (df["City"] == "Calgary") & (df["Year"] == 1988)][["Name", "Age", "NOC", "Sport"]].sort_values(by="Age", ascending=False).drop_duplicates().head(5)
res2
```

```python
#Pregunta 3
res3 = df[(df["City"] == "Montreal") & (df["Year"] == 1976) & (df["Sport"] == "Gymnastics")][["Event"]].drop_duplicates()
display(res3)
res3.count()
```

```python
#Pregunta 4
res4 = df[(df["City"] == "Atlanta") & (df["Year"] == 1996) & (df["Sport"] == "Cycling")]
res4.groupby("Team")[["ID"]].nunique().sort_values(by="ID", ascending=False).head(5)
```

```python
#Pregunta 5
res5 = df[(df["Season"] == "Winter") & (df["City"] == "Lillehammer") & (df["Year"] == 1994)]
res5p1 = res5.groupby("Sport")[["Age"]].mean().sort_values(by="Age", ascending=False).head(1)
res5p2 = res5.groupby("Sport")[["Age"]].mean().sort_values(by="Age", ascending=True).head(1)
display(res5p1)
display(res5p2)

```

```python
#Pregunta 6
copia = df.copy()

copia["Diferencia"] = copia["Height"] - copia["Weight"]
res6 = copia[(copia["City"] == "Mexico City") & (copia["Year"] == 1968)][["Name", "Sport", "Height", "Weight", "Diferencia"]].sort_values(by="Weight", ascending=False).head(5)
res6
```

```python
#Pregunta 7
copia2 = df.copy()
res7 = copia2[(copia2["Season"] == "Summer") & ((copia2["Year"] == 1964) | (copia2["Year"] == 1968) | (copia2["Year"] == 1972)) & ((copia2["NOC"] == "USA") | (copia2["NOC"] == "URS") | (copia2["NOC"] == "JPN"))][["Year", "NOC", "Medal"]]
res7.pivot_table(index="Medal", columns="NOC", aggfunc="count")
```
