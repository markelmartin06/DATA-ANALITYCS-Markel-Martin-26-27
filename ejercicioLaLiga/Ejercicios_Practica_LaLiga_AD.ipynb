## **Ejercicio 3 – Agrupaciones (`groupby`)**
Trabaja con agregaciones por equipo y posición:

1. Calcula el **total de goles anotados por cada equipo** usando `Goals scored`.
2. Calcula el **total de asistencias por equipo** (`Assists`).
3. Calcula la **media de minutos jugados por posición** (`Position`, `Minutes played`).
4. Para cada equipo (`Team`), muestra el **total de tarjetas amarillas** (`Yellow Cards`) y **tarjetas rojas** (`Red Cards`).
5. Calcula el **número total de duelos ganados** (`Successful duels`) por equipo.
6. Para cada posición (`Position`), calcula la **media de tiros a puerta** (`Shots on target`).


```python
# Tu código aquí
ej3rs1 = laliga.groupby("Team")["Goals scored"].sum()
display(ej3rs1)
ej3rs2 = laliga.groupby("Team")["Assists"].sum()
display(ej3rs2)
ej3rs3 = laliga.groupby("Position")["Minutes played"].mean()
display(ej3rs3)
ej3rs4P1 = laliga.groupby("Team")["Red Cards"].sum()
ej3rs4P2 = laliga.groupby("Team")["Yellow Cards"].sum()
display(ej3rs4P1)
display(ej3rs4P2)
ej3rs5 = laliga.groupby("Team")["Successful duels"].sum()
display(ej3rs5)
ej3rs6 = laliga.groupby("Position")["Shots on target"].mean()
display(ej3rs6)
```

## **Ejercicio 6 – Segmentación (`pd.cut`)**
Usa `pd.cut` para crear categorías a partir de columnas numéricas:

1. Crea una nueva columna `Rango_minutos` a partir de `Minutes played` con estas categorías:
   - `Poco`: menos de 500 minutos
   - `Medio`: entre 500 y 1500 minutos
   - `Mucho`: más de 1500 minutos
   Después, muestra cuántos jugadores hay en cada categoría (`value_counts`).

2. Crea una columna `Rango_goles` a partir de `Goals scored` con estas categorías:
   - `Sin gol`: 0 goles
   - `Pocos`: 1–3 goles
   - `Goleadores`: 4 o más goles
   Muestra también cuántos jugadores hay en cada categoría.

3. Crea una columna `Rango_tiros` a partir de `Shots` con intervalos que tú elijas (por ejemplo: 0–10, 11–30, 31+). Muestra la distribución con `value_counts()`.
4. Muestra ejemplos (5 filas) de cada una de las categorías de `Rango_minutos` y `Rango_goles`.


```python
# Tu código aquí
laliga["Minutos_reales"] = laliga["Minutes played"].mask(laliga["Minutes played"] != laliga["Minutes played"].round(), laliga["Minutes played"] * 1000).round()
laliga["Rango_minutos"] = pd.cut(laliga["Minutos_reales"], bins=[-1, 499, 1500, 10000], labels=["Poco", "Medio", "Mucho"])
ej6rs1 = laliga["Rango_minutos"].value_counts()
display(ej6rs1)
laliga["Rango_goles"] = pd.cut(laliga["Goals scored"], bins=[-1, 0, 3, 100], labels=["Sin gol", "Pocos", "Goleadores"])
ej6rs2 = laliga["Rango_goles"].value_counts()
display(ej6rs2)
laliga["Rango_tiros"] = pd.cut(laliga["Shots"], bins=[-1, 10, 30, 1000], labels=["0-10", "11-30", "31+"])
ej6rs3 = laliga["Rango_tiros"].value_counts()
display(ej6rs3)
ej6rs4P1 = laliga[laliga["Rango_minutos"] == "Poco"][["Name", "Minutos_reales", "Rango_minutos"]].head(5)
display(ej6rs4P1)
ej6rs4P2 = laliga[laliga["Rango_minutos"] == "Medio"][["Name", "Minutos_reales", "Rango_minutos"]].head(5)
display(ej6rs4P2)
ej6rs4P3 = laliga[laliga["Rango_minutos"] == "Mucho"][["Name", "Minutos_reales", "Rango_minutos"]].head(5)
display(ej6rs4P3)
ej6rs4P4 = laliga[laliga["Rango_goles"] == "Sin gol"][["Name", "Goals scored", "Rango_goles"]].head(5)
display(ej6rs4P4)
ej6rs4P5 = laliga[laliga["Rango_goles"] == "Pocos"][["Name", "Goals scored", "Rango_goles"]].head(5)
display(ej6rs4P5)
ej6rs4P6 = laliga[laliga["Rango_goles"] == "Goleadores"][["Name", "Goals scored", "Rango_goles"]].head(5)
display(ej6rs4P6)
```
